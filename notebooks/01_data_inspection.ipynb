### Outliers Detection Using IQR

Using the Interquartile Range (IQR) method
to identify potential outliers.


```python
Q1 = df[temperature_columns].quantile(0.25)
Q3 = df[temperature_columns].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = (
    (df[temperature_columns] < lower_bound) |
    (df[temperature_columns] > upper_bound)
).sum()

outliers
```

```python
# Prepare annual temperature data

annual_df = df.dropna(subset=["J-D"]).copy()

# Display the annual dataset
annual_df[["Year", "J-D"]].tail(10)
```

```python
# Statistical analysis of annual temperature anomalies

annual_df["J-D"].describe()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(14, 6))

plt.plot(
    annual_df["Year"],
    annual_df["J-D"],
    label="Annual Temperature Anomaly"
)

plt.axhline(y=0, linestyle="--", label="Reference Level")

plt.title("Global Annual Temperature Anomaly (1880–2025)")
plt.xlabel("Year")
plt.ylabel("Temperature Anomaly (°C)")
plt.legend()
plt.grid(True)

plt.show()
```

```python
# Calculate 10-year moving average

annual_df["MA_10"] = (
    annual_df["J-D"].rolling(window=10).mean()
)

# Plot annual values and moving average

plt.figure(figsize=(14, 6))

plt.plot(
    annual_df["Year"],
    annual_df["J-D"],
    alpha=0.4,
    label="Annual Anomaly"
)

plt.plot(
    annual_df["Year"],
    annual_df["MA_10"],
    linewidth=2,
    label="10-Year Moving Average"
)

plt.axhline(y=0, linestyle="--", label="Reference Level")

plt.title("Global Temperature Anomaly – 10-Year Moving Average")
plt.xlabel("Year")
plt.ylabel("Temperature Anomaly (°C)")
plt.legend()
plt.grid(True)

plt.show()
```

### EDA – Annual Temperature Trend Analysis

A 10-year moving average was calculated to identify the long-term temperature trend and reduce short-term annual fluctuations.

**Key Findings:**

- The early historical period shows predominantly negative temperature anomalies.
- A gradual warming trend appears during the early 20th century.
- Annual anomalies fluctuate around the reference level during the middle of the century.
- A clear long-term warming trend emerges from approximately the 1970s.
- The moving average increases substantially during the most recent decades.

**Conclusion:**

The 10-year moving average reveals a strong long-term warming trend in global surface temperature anomalies.

The moving average smooths short-term fluctuations without modifying the original annual observations.

```python
# Group annual temperature data by decade

annual_df["Decade"] = (annual_df["Year"] // 10) * 10

decadal_df = annual_df.groupby("Decade").agg(
    Mean_Temperature=("J-D", "mean"),
    Number_of_Years=("Year", "count")
).reset_index()

# Display decadal results
decadal_df
```

```python
# Visualize decadal temperature anomalies

plt.figure(figsize=(14, 6))

plt.bar(
    decadal_df["Decade"].astype(str),
    decadal_df["Mean_Temperature"]
)

plt.axhline(y=0, linestyle="--")

plt.title("Average Global Temperature Anomaly by Decade")
plt.xlabel("Decade")
plt.ylabel("Mean Temperature Anomaly (°C)")

plt.xticks(rotation=45)
plt.grid(axis="y", alpha=0.3)

plt.show()
```

## Decadal Temperature Analysis – Key Findings

The decadal analysis reveals a clear long-term warming trend
in global temperature anomalies.

### Key Findings

```python
# Prepare valid annual temperature data

trend_df = annual_df.dropna(subset=["J-D"]).copy()

# Get first and last available years
first_year = trend_df.iloc[0]
last_year = trend_df.iloc[-1]

# Calculate total temperature anomaly change
temperature_change = (
    last_year["J-D"] - first_year["J-D"]
)

print("First year:", int(first_year["Year"]))
print("First anomaly:", first_year["J-D"])

print("Last year:", int(last_year["Year"]))
print("Last anomaly:", last_year["J-D"])

print("Total change:", round(temperature_change, 3), "°C")
```

```python
# Calculate the long-term linear temperature trend

slope, intercept = np.polyfit(
    trend_df["Year"],
    trend_df["J-D"],
    1
)

trend_df["Trend"] = (
    slope * trend_df["Year"] + intercept
)

print("Temperature trend:", round(slope, 4), "°C/year")

print("Temperature trend per decade:",
      round(slope * 10, 3), "°C/decade")
```

```python
# Visualize long-term temperature trend

plt.figure(figsize=(14, 6))

plt.plot(
    trend_df["Year"],
    trend_df["J-D"],
    label="Annual Temperature Anomaly"
)

plt.plot(
    trend_df["Year"],
    trend_df["Trend"],
    linestyle="--",
    label="Linear Trend"
)

plt.axhline(y=0, linestyle=":", alpha=0.7)

plt.title("Long-Term Global Temperature Anomaly Trend (1880–2025)")
plt.xlabel("Year")
plt.ylabel("Temperature Anomaly (°C)")

plt.legend()
plt.grid(alpha=0.3)

plt.show()
```
