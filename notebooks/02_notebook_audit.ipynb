```python

import json
from pathlib import Path
from collections import Counter

# Locate the original notebook
notebook_path = Path("01_data_inspection.ipynb")

with open(notebook_path, "r", encoding="utf-8") as file:
    notebook = json.load(file)

cells = notebook["cells"]

print("=== NOTEBOOK AUDIT ===")
print("Total cells:", len(cells))

# Count cell types
cell_types = Counter(cell["cell_type"] for cell in cells)

print("\nCell types:")
for cell_type, count in cell_types.items():
    print(f"{cell_type}: {count}")

# Display Markdown headings
print("\n=== MARKDOWN HEADINGS ===")

for i, cell in enumerate(cells, start=1):
    if cell["cell_type"] == "markdown":
        source = "".join(cell["source"]).strip()

        if source.startswith("#"):
            print(f"Cell {i}: {source.splitlines()[0]}")

# Detect exact duplicate cells
print("\n=== EXACT DUPLICATES ===")

seen = {}
duplicates = []

for i, cell in enumerate(cells, start=1):
    source = "".join(cell["source"]).strip()

    if not source:
        continue

    key = (cell["cell_type"], source)

    if key in seen:
        duplicates.append((seen[key], i))
    else:
        seen[key] = i

if duplicates:
    for first, second in duplicates:
        print(f"Cell {first} and Cell {second}")
else:
    print("No exact duplicate cells found.")

# Detect empty cells
print("\n=== EMPTY CELLS ===")

empty_cells = [
    i for i, cell in enumerate(cells, start=1)
    if not "".join(cell["source"]).strip()
]

print(empty_cells if empty_cells else "No empty cells.")
```

```python

# Inspect exact duplicate cells

duplicate_pairs = [
    (3, 4),
    (10, 17),
    (9, 19),
    (14, 20),
    (18, 25),
    (28, 29),
    (9, 34),
    (41, 43),
    (10, 45),
    (18, 46),
    (12, 49),
    (9, 50),
    (3, 112),
    (37, 117)
]

for first, second in duplicate_pairs:
    print("\n" + "=" * 60)
    print(f"CELL {first} vs CELL {second}")

    for number in [first, second]:
        cell = cells[number - 1]

        print(f"\n--- CELL {number} ({cell['cell_type']}) ---")
        print("".join(cell["source"]))
```

```python

# Inspect context around duplicated cells

duplicate_groups = [
    [3, 4, 112],
    [9, 19, 34, 50],
    [10, 17, 45],
    [12, 49],
    [14, 20],
    [18, 25, 46],
    [28, 29],
    [37, 117],
    [41, 43]
]

for group in duplicate_groups:
    print("\n" + "=" * 65)
    print("DUPLICATE GROUP:", group)

    for number in group:
        print(f"\n--- CELL {number} ---")

        start = max(1, number - 2)
        end = min(len(cells), number + 2)

        for i in range(start, end + 1):
            cell = cells[i - 1]
            source = "".join(cell["source"]).strip()
            preview = source[:120].replace("\n", " ")

            print(
                f"Cell {i} | {cell['cell_type']} | {preview}"
            )
```

```python

import json
from pathlib import Path

# Original notebook
source_path = Path("01_data_inspection.ipynb")

# New cleaned notebook
output_path = Path("01_data_inspection_cleaned.ipynb")

# Load original notebook
with open(source_path, "r", encoding="utf-8") as file:
    nb = json.load(file)

# Cells confirmed as redundant (original numbering)
remove_cells = {4, 29, 43, 57, 122}

# Create a cleaned copy
nb["cells"] = [
    cell
    for index, cell in enumerate(nb["cells"], start=1)
    if index not in remove_cells
]

# Save without modifying the original
with open(output_path, "w", encoding="utf-8") as file:
    json.dump(nb, file, ensure_ascii=False, indent=1)

print("Cleaning copy created successfully!")
print("Original cells:", 122)
print("Removed cells:", len(remove_cells))
print("Cleaned cells:", len(nb["cells"]))
print("Output:", output_path)
```

```python

import nbformat
from pathlib import Path
from collections import defaultdict

# Notebook to audit
notebook_path = Path("../notebooks/01_data_inspection_cleaned.ipynb")

nb = nbformat.read(notebook_path, as_version=4)

duplicates = defaultdict(list)

for index, cell in enumerate(nb.cells, start=1):
    if cell.cell_type == "code" and cell.source.strip():
        code = cell.source.strip()
        duplicates[code].append(index)

print("=== DUPLICATE DETECTION REPORT ===")

found = False

for code, indexes in duplicates.items():
    if len(indexes) > 1:
        found = True
        print(f"\nDuplicate cells: {indexes}")
        print("Code preview:")
        print(code[:300])
        print("-" * 60)

if not found:
    print("No exact duplicate code cells found.")

print("\nTotal cells:", len(nb.cells))
```

```python

import nbformat
from pathlib import Path

source_path = Path("../notebooks/01_data_inspection_cleaned.ipynb")
output_path = Path("../notebooks/01_data_inspection_cleaned_v2.ipynb")

# Load cleaned notebook
notebook = nbformat.read(source_path, as_version=4)

# Confirmed duplicate cells to remove
cells_to_remove = {108, 18, 32, 47, 16, 42, 46, 19, 24, 43}

# Create a new notebook without modifying the source
new_notebook = nbformat.v4.new_notebook()
new_notebook.metadata = notebook.metadata.copy()

new_notebook.cells = [
    cell for index, cell in enumerate(notebook.cells, start=1)
    if index not in cells_to_remove
]

# Save new version
nbformat.write(new_notebook, output_path)

print("Automatic cleaning completed!")
print("Original cells:", len(notebook.cells))
print("Removed cells:", len(cells_to_remove))
print("New cells:", len(new_notebook.cells))
print("Output:", output_path)
```

```python

import nbformat
from pathlib import Path

# Path to the cleaned v2 notebook
path = Path("../notebooks/01_data_inspection_cleaned_v2.ipynb")

# Load notebook
nb = nbformat.read(path, as_version=4)

# Validate notebook structure
nbformat.validate(nb)

print("=" * 50)
print("NOTEBOOK VALIDATION: PASSED")
print("=" * 50)

print("Total cells:", len(nb.cells))
print("Code cells:", sum(c.cell_type == "code" for c in nb.cells))
print("Markdown cells:", sum(c.cell_type == "markdown" for c in nb.cells))

# Check empty cells
empty_cells = [
    i for i, cell in enumerate(nb.cells, start=1)
    if not cell.source.strip()
]

print("\nEmpty cells:", empty_cells)
print("Number of empty cells:", len(empty_cells))

# Check code cells with stored execution errors
error_cells = [
    i for i, cell in enumerate(nb.cells, start=1)
    if cell.cell_type == "code"
    and any(output.output_type == "error" for output in cell.get("outputs", []))
]

print("\nCells with stored execution errors:", error_cells)
print("Number of error cells:", len(error_cells))
```

```python

import nbformat
from pathlib import Path

path = Path("../notebooks/01_data_inspection_cleaned_v2.ipynb")

nb = nbformat.read(path, as_version=4)

removed_errors = 0

for cell in nb.cells:
    if cell.cell_type == "code":
        outputs = cell.get("outputs", [])

        clean_outputs = []

        for output in outputs:
            if output.output_type == "error":
                removed_errors += 1
            else:
                clean_outputs.append(output)

        cell.outputs = clean_outputs

nbformat.validate(nb)

nbformat.write(nb, path)

print("Stored errors removed:", removed_errors)
print("Notebook validation: PASSED")
print("Total cells:", len(nb.cells))
```
